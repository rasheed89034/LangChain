#### Faiss
Facebook AI Similarity Search (Faiss) is a library for efficient similarity search and clustering of dense vectors. It contains algorithms that search in sets of vectors of any size, up to ones that possibly do not fit in RAM. It also contains supporting code for evaluation and parameter tuning.

```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_classic.embeddings import OllamaEmbeddings 
from langchain_text_splitters import CharacterTextSplitter
```

```python
loader = TextLoader("speech.txt")
documents = loader.load()
textSplitter = CharacterTextSplitter(chunk_size = 500,chunk_overlap = 50)
docs = textSplitter.split_documents(documents)
docs
```

```python
from PIL import GimpGradientFile
embeddings = OllamaEmbeddings(model="embeddinggemma")
faissDB = FAISS.from_documents(documents=docs,embedding=embeddings)
faissDB
```

```python
query = "Describe the summary of the speech"
docs = faissDB.similarity_search(query)
docs[0].page_content
```

```python
retirever = faissDB.as_retriever()
answer = retirever.invoke(query)
answer
```

```python
docsAndScore = faissDB.similarity_search_with_score(query)
docsAndScore
```

```python
## Store in your local disk
faissDB.save_local("faiss_index")
```

```python
openDB = FAISS.load_local("faiss_index",embeddings,allow_dangerous_deserialization=True)
openDB
```

```python
## The answer should be same
querySearchFromLocalDB = openDB.similarity_search(query)
querySearchFromLocalDB
```
